```python
from polyglot.text import Word
```

```python
words=['cats','computing','association','programming','leadership','miscommunication','identifiable',
      'psychologically','engineering','tabular','realistic','colorfulness']
```

```python
for w in words:
    w=Word(w,language='en')
    print(w,'\t',w.morphemes)
```

```python
mwords=['राज्यशास्त्र','जलविदुयत' ,'महाद्वीप' ]
for word in mwords:
    w= Word(word,language='mr')
    print(word,'\t\t',w.morphemes)
```
