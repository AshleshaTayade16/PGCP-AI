```python
from nltk import TnT
```

```python
#create the object 
tagger =TnT()
```

```python
#extract the tags
tags=indian.tagged_sents('marathi.pos')
```

```python
# train the tagger
tagger.train(tags)
```

```python
new_sent='हा सामना सोमवारी उशिरा  सुरु झाला.'
```

```python
tagger.tag(word_tokenize(new_sent))
```
