# Question answering

```python
qa = pipeline('question-answering')
```

```python
text = 'New Delhi is capital city of India'
question = 'What is capital city of India?'
result = qa(question=question, context=text)
df = pd.DataFrame([result])
df
```

# Translation

```python
trans = pipeline('translation_en_to_fr')
```

```python
translator = pipeline(
    'translation',
    model='facebook/m2m100_418M',
    src_lang='en',
    tgt_lang='mr'
)

text = "The quick brown fox jumps over the lazy dog"
# text='where is Sai? did she ate apple'
result = translator(text)

# Changed 'translated_text' to 'translation_text'
print("Translated text: ", result[0]['translation_text'])
```

```python
text = "Shraddha is a good girl"
result = translator(text)
print("Translated text: ", result[0]['translation_text'])
```
