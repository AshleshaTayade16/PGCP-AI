```python
### Question 1
# Download the data from the link below using web scrapping:https://en.wikipedia.org/wiki/Mount_Everest
# Find the count of Total “Mount” in the text.
# Save the extracted plain text contents in file mount.txt in upper case.
# Print the title of this page.
```

```python
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://en.wikipedia.org/wiki/Mount_Everest"
headers = {'User-Agent': 'Mozilla/5.0'}
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')
```

```python
plain_text = soup.get_text()
plain_text = plain_text.replace('\n', '').replace('\t', ' ')
print(plain_text)
```

```python
# count of Total “Mount” in the text.
mount_count = plain_text.count("Mount")
print(f"Total count of 'Mount': {mount_count}")
```

```python
with open("mount.txt", "w", encoding="utf-8") as file:
    file.write(plain_text.upper())

with open("mount.txt", 'r', encoding='utf-8') as f:
    print(f.read())
```

```python
# TITLE
page_title = soup.title.string if soup.title else "No title found"
print(f"Page Title: {page_title}")
```

```python
from pypdf import PdfReader

pdf = "D:\\AshleshaRuchika\\PGCP-AI\\NLP\\Assignments\\introduction-to-natural-language-processing.pdf"
```

```python
combined_text = ""
reader = PdfReader(pdf)
# text from page 12 and 13
page_12 = reader.pages[11].extract_text()
page_13 = reader.pages[12].extract_text()

if page_12:
    combined_text += page_12 + "\n\n"
if page_13:
    combined_text += page_13 + "\n\n"

combined_text
```

```python
### Question 3
# Read the file from the link givcen below:
# https://icml.cc/Conferences/2013/wp-content/uploads/2013/06/Machine-Learning-and-Natural-Language-Processing.docx
# Read and display first 5 paragraphs from it.
# Count total ‘The’ present in the text.
```

```python
import docx
```

```python
# Load the Word document
doc_path = "D:\\AshleshaRuchika\\PGCP-AI\\NLP\\Assignments\\Machine-Learning-and-Natural-Language-Processing.docx"
doc = docx.Document(doc_path)
```

```python
# First 5 Paragraphs
paragraph_count = 0
full_text = ""

for para in doc.paragraphs:
    full_text += para.text + "\n"
    
    if paragraph_count < 5:
        print(para.text)
        print()
        paragraph_count += 1
```

```python
# total ‘The’ present in the text
words = full_text.split()
the_count = 0

for word in words:
    cleaned_word = word.strip(".,!?()\"':;")
    if cleaned_word == 'The':
        the_count += 1

print(f"Total 'The' (case-sensitive) present in the text: {the_count}")
```
