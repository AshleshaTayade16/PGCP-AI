```python
import builtins

# Save the original open function
original_open = builtins.open

# Create a wrapper that forces utf-8 encoding by default
def utf8_open(*args, **kwargs):
    if 'encoding' not in kwargs:
        kwargs['encoding'] = 'utf-8'
    return original_open(*args, **kwargs)

# Override the built-in open function
builtins.open = utf8_open

# Now re-run your IndoWordNet code
import pyiwn
iwn = pyiwn.IndoWordNet(lang=pyiwn.Language.MARATHI)
```

```python
iwn= pyiwn.IndoWordNet(lang=pyiwn.Language.MARATHI)
```

```python
word=iwn.synsets('आश्लेषा')
```

```python
hyponyms = iwn.synset_relation(word[0], pyiwn.SynsetRelations.HYPONYMY)
```

```python
for h in hyponyms:
    print(h.lemma_names()[0])
```
