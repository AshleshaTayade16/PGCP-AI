## Read all the pdf files,extrac their contents and store in file data.txt

```python

reader = PdfReader("D:\\AshleshaRuchika\\PGCP-AI\\NLP\\gunesh.pdf")
```

```python
text = ''
for file in os.listdir(reader):
    if file.endswith('.pdf'):
        reader = PdfReader(reader+file)
        for page in reader.pages:
            text += page.extract_text()

with open('data.txt', 'w') as f:
    f.write(text)
```

```python
import os
from pypdf import PdfReader

# 1. Define your base directory clearly
base_dir = "D:\\AshleshaRuchika\\PGCP-AI\\NLP\\"

text = ''

# 2. Iterate through the target directory
for file in os.listdir(base_dir):
    if file.endswith('.pdf'):
        # 3. Combine the directory path with the filename safely
        full_path = os.path.join(base_dir, file)
        
        reader = PdfReader(full_path)
        for page in reader.pages:
            text += page.extract_text()

# 4. Save the combined text to your file
with open('data.txt', 'w', encoding='utf-8') as f:
    f.write(text)
```
