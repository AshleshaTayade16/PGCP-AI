### Find top 10 words occured frequnetly in this article

```python
from nltk.tokenize import word_tokenize
```

```python
tokens= word_tokenize(text)
```

```python
from nltk import FreqDist
```

```python
freq=FreqDist(tokens)
```

```python
freq.plot(10)
```

```python
from nltk.corpus import stopwords
from nltk.stem import LancasterStemmer
ls=LancasterStemmer()
```

```python
token1=[ls.stem(token.lower()) for token in tokens
       if token.isalpha() and token.lower() 
        not in stopwords.words('english')]
```

```python
freq=FreqDist(token1)
freq.plot(20)
```

```python
freq.most_common(20)
```

```python
swords=stopwords.words('english') + ['retrived', 'archived','edit','b','original'],
```

```python
url = 'https://en.wikipedia.org/wiki/Saturn'
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')
text = soup.get_text()
tokens = word_tokenize(text)
tokens1 = [token.lower() for token in tokens
           if token.isalpha() and token.lower()
           not in swords]
freq = FreqDist(tokens1)
freq.plot(20)
```
