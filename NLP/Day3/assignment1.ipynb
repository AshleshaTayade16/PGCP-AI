```python
cleaned_chars = []
punctuation_list = []

for char in sent:
    if char.isalnum() or char.isspace():
        cleaned_chars.append(char)
    else:
        # If it's not a letter, number, or space, it's a punctuation mark
        punctuation_list.append(char)
```

```python
clean_text = "".join(cleaned_chars)
tokens = clean_text.split()
print(f"Processed Tokens: {tokens}\n")
```

```python
# 1. Find the average of ages.
ages = [int(t) for t in tokens if t.isdigit()]
avg_age = sum(ages) / len(ages) if ages else 0
print(f"1. Average of ages: {avg_age:.2f}")
```

```python
# 2. Count total number of words starting with 't' / 'T'
words_t = [t for t in tokens if t.lower().startswith('t')]
print(f"2. Total words starting with 't'/'T': {len(words_t)} -> {words_t}")
```

```python
# 3. Count total punctuation symbols.
total_punctuations = len(punctuation_list)
print(f"3. Total punctuation symbols: {total_punctuations} -> {punctuation_list}")
```

```python
# 4. Print the words in capital letter which ends with 'd'
# .isalpha() ensures we only look at actual words, not numbers
words_d = [t.upper() for t in tokens if t.lower().endswith('d') and t.isalpha()]
print(f"4. Words ending with 'd' in uppercase: {words_d}")
```

```python
# 5. Merge all the words and print (no punctuations)
merged_no_space = "".join(tokens)
merged_with_space = " ".join(tokens)
print(f"5. Merged words (without space): {merged_no_space}")
print(f"   Merged words (with space):    {merged_with_space}")
```
