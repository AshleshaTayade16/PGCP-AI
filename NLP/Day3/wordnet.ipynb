# Extract the antonyms

```python
antonyms = set()

for word in wordnet.synsets('possible'):
    for lemma in word.lemmas():
        if lemma.antonyms():
            antonyms.add(lemma.antonyms()[0].name())
```

```python
antonyms
```

```python
antonyms = set()

for word in wordnet.synsets('computer'):
    for lemma in word.lemmas():
        if lemma.antonyms():
            antonyms.add(lemma.antonyms()[0].name())
```

```python
antonyms
```

```python
word = wordnet.synsets('intelligence')
for x in word:
    hypernyms= x.hypernyms()
    for h in hypernyms:
        print(h.name())
```

```python
word = wordnet.synsets('dog')
for x in word:
    hypernyms= x.hypernyms()
    for h in hypernyms:
        print(h.name())
```

## extarct meronyms

```python
word = wordnet.synsets('computer')
for x in word:
    meronyms= x.part_meronyms()
    for m in meronyms:
        print(m.name())
```
