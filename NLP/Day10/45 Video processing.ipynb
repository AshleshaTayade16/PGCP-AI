```python
# Extract all the frames and store in folder /nature

Vid = cv2.VideoCapture('D:\\AshleshaRuchika\\PGCP-AI\\NLP\\images\\nature.mp4')
i = 1
while True:
    ret, frame = Vid.read()
    if ret == False:
        break
    cv2.imwrite('D:\\AshleshaRuchika\\PGCP-AI\\NLP\\Day10\\nature\\image%d.jpg' % i, frame)
    i += 1

cv2.destroyAllWindows()
print('All frames saved!')
```

```python
vid = cv2.VideoCapture(0)

frame_width = int(vid.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(vid.get(cv2.CAP_PROP_FRAME_HEIGHT))
fourcc = cv2.VideoWriter_fourcc(*'XVID')
out = cv2.VideoWriter('output.avi', fourcc, 30.0, (frame_width, frame_height))

while True:
    ret, frame = vid.read()
    out.write(frame)
    cv2.imshow('My Video', frame)
    if cv2.waitKey(5) == ord('q'):
        break

vid.release()
out.release()
cv2.destroyAllWindows()
```
