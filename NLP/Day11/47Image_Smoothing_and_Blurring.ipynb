```python
img=cv2.imread("D:\\AshleshaRuchika\\PGCP-AI\\NLP\\images\\cat_damaged.png")
cv2.imshow("Image",img)
```

```python
kernel=np.zeros([3,3])+1/9
```

```python
kernel
```

```python
blurred_img=cv2.filter2D(img,-1,kernel)
```

```python
cv2.imshow('Blurred Image', blurred_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
kernel=np.zeros([9,9])+1/81
blurred_img=cv2.filter2D(img,-1,kernel)
img_new=np.hstack([img,blurred_img])
cv2.imshow('Blurred Image',img_new)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Gaussian Kernel

```python
kernel_size=7
kernel=cv2.getGaussianKernel(kernel_size,1.23)
```

```python
kernel
```

```python
gauss_image=cv2.filter2D(img,-1,kernel)
img_new=np.hstack([img,gauss_image])
cv2.imshow('Blurred Image', img_new)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Median Blue

```python
img_gray=cv2.imread("D:\\AshleshaRuchika\\PGCP-AI\\NLP\\images\\Dog2.png")
cv2.imshow("Dog",img_gray)
```

```python
img_gray.shape
```

```python
img_med=cv2.medianBlur(img_gray,7)
```

```python
plt.figure(figsize=(16,9))
plt.subplot(121)
plt.imshow(img_gray,cmap='gray')
plt.subplot(122)
plt.imshow(img_med,cmap='gray')
```

```python
img_med=cv2.medianBlur(img_gray[100:251,100:251],7)
img_copy=img_gray.copy()
img_copy[100:251,100:251]=img_med
```

```python
plt.figure(figsize=(16,9))
plt.subplot(121)
plt.imshow(img_gray,cmap='gray')
plt.subplot(122)
plt.imshow(img_copy,cmap='gray')
```
