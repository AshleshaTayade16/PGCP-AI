```python
lines=[
    'It was a nice rainy day.',
    'The things were so beautiful on his point.',
    'When your focus is clear, you won.',
    'Many may happy returns of the day!',
    'One book can change your life.',
]
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
tokenizer=Tokenizer()
tokenizer.fit_on_texts(lines)
```

```python
tokenizer.word_docs
```

```python
tokenizer.word_index
```

```python
tokenizer.index_word
```

```python
tokenizer.index_docs
```

```python
mat=tokenizer.texts_to_matrix(lines)
```

```python
mat
```

```python
mat.shape
```

```python
for i, embeddings in enumerate(mat[0]):
    if embeddings==1.0 and i!=0:
        print(tokenizer.index_word[i])
```

```python
words_in_first_row = [
    tokenizer.index_word[i] for i, value in enumerate(mat[1])
    if value == 1.0 and i != 0
]
```

```python
words_in_first_row
```

```python
seq =  tokenizer.texts_to_sequences(lines)
```

```python
seq
```

```python
padded = pad_sequences(seq,maxlen = 10 , padding= "post")
padded
```

```python
padded = pad_sequences(seq,maxlen = 10 , padding= "pre")  ### very use method
padded
```

```python
max_length = max(len(s) for s in seq)
print(max_length)
```

```python
padded=pad_sequences(seq, maxlen=max([len(i) for i in seq]), padding='pre')
```
