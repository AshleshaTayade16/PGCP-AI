```python
alphabets = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'
```

```python
len(alphabets)
```

```python
list(enumerate(alphabets))
```

```python
int_to_char = dict(enumerate(alphabets))
```

```python
char_to_int = dict((v,i) for i,v in enumerate(alphabets))
```

```python
char_to_int
```

#### prepare the data

```python
seq_length = 1
X = []
y = []

for i in range(len(alphabets) - seq_length):
    seq_in = alphabets[i]
    seq_out = alphabets[i+1]
    print(seq_in, '-->', seq_out)
    X+=([char_to_int[seq_in[0]]])
    y+=([char_to_int[seq_out]])
```

```python
X
```

```python
y
```

```python
import numpy as np
```

```python
X= np.reshape(X,(25,1,1))
```

```python
X
```

```python
X_scaled= X/25
```

```python
X_scaled
```

```python
from keras.utils import to_categorical
```

```python
y_new = to_categorical(y)
```

```python
y_new
```

```python
y_new.shape
```

## build the model

```python
from keras.layers import SimpleRNN ,Input, Dense
from keras.models import Sequential
```

```python
model = Sequential()
model.add(Input((1,1)))
model.add(Dense(32,activation="relu"))
model.add(SimpleRNN(32))
model.add(Dense(26,activation="softmax"))
```

```python
model.summary()
```

```python
model.compile(loss="categorical_crossentropy", optimizer="adam", metrics=["accuracy"])
```

```python
model.fit(X_scaled,y_new,batch_size=1,epochs=500)
```

```python
model.evaluate(X_scaled, y_new,batch_size=1)
```

```python
new_char='K'
new_char=np.reshape(char_to_int[new_char],(1,1,1))
new_char=new_char/25
y_pred=model.predict(new_char,verbose=0)
int_to_char[y_pred.argmax()]
```

```python
seq_length = 3
x = []
y = []

for i in range(len(alphabets) - seq_length):
  seq_in = [alphabets[i:i+seq_length]]
  seq_out = alphabets[i+seq_length]
  print(seq_in, '--->',seq_out)
  x.append([char_to_int[char] for char in seq_in[0]])
  y +=[char_to_int[seq_out]]
```

```python
x_new = np.reshape(x,(23,seq_length,1))
```

```python
y_new = to_categorical(y)
```

```python
x_new.shape,y_new.shape
```

```python
x_new = x_new/25
```

```python
model = Sequential()

model.add(Input((3,1)))
model.add(Dense(32,activation='relu'))
model.add(SimpleRNN(32))
model.add(Dense(26,activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy',
              optimizer='adam',metrics=['accuracy'])
```

```python
model.fit(x_new,y_new,batch_size=1,epochs=1000)
```

```python
new='JKL'
new_char = np.array([[char_to_int[x]] for x in new])
new_char = np.reshape(new_char,(1,3,1))
new_char = new_char/25
pred = model.predict(new_char,verbose=0)
int_to_char[np.argmax(pred)]
```
