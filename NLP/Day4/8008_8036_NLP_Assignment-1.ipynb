```python
import string
import requests
from bs4 import BeautifulSoup
```

```python
'''Q.1
sent = 'They told that their ages are 23, 25 and 29 respectively.'
# 1. Find the average of ages.
# 2. Count total number of words starts with 't' / 'T'
# 3. Count total punctuation symbols.
# 4. Print the words in capital letter which ends with 'd'
# 5. Merge all the words and print (no punctuations)'''
```

```python
sent = 'They told that their ages are 23, 25 and 29 respectively.'
```

#### 1. Find the average of ages.

```python
ages = [int(word.strip(string.punctuation)) for word in sent.split() if word.strip(string.punctuation).isdigit()]
print(ages)
print(f"Avg: {sum(ages)/len(ages):.2f}")
```

####  2. Count total number of words starts with 't' / 'T'

```python
words = sent.split()
words_starting_with_t = [word for word in words if word.lower().startswith('t')]
print("Total words starting with 't'/'T':", len(words_starting_with_t))
```

#### 3. Count total punctuation symbols.

```python
punctuation_count = 0
for char in sent:
    if char in string.punctuation:
        punctuation_count += 1

print("Total punctuation symbols:", punctuation_count)
```

#### 4. Print the words in capital letter which ends with 'd'

```python
words_ending_with_d = [word for word in words if word.lower().endswith('d')]
print("Words ending with 'd' in capital letters:", words_ending_with_d)
```

#### 5. Merge all the words and print (no punctuations)

```python
clean_tokens = [word.strip(string.punctuation) for word in words if word.strip(string.punctuation)]
merged_sent = " ".join(clean_tokens)
print(f"Merged string: {merged_sent}")
```

```python
'''Q.2
# Open the wikipedia page of Indian flag and perform following analysis
# Count total number of Words
# Count percentage of punctuation marks
# Print total number of lines
# How many words are in title case?
# Remove the numbers and print the text'''
```

#### Fetch the Wikipedia page for the National Flag of India

```python
url = "https://en.wikipedia.org/wiki/Flag_of_India"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')
```

#### Extract raw text lines using a basic for loop

```python
text_lines = [p.get_text().strip() for p in soup.find_all('p') if p.get_text().strip()]
```

#### Combine lines into one string and break down into single words

```python
# Combine lines into one main text block and break it down into words
full_text = "\n".join(text_lines)
all_words = full_text.split()
```

#### Count total number of Words

```python
total_words = len(all_words)
print(f"Total number of words: {total_words}")
```

####  Count percentage of punctuation marks




```python
punctuation_chars = sum(1 for char in full_text if char in string.punctuation)
total_chars = len(full_text)
punctuation_percentage = (punctuation_chars / total_chars) * 100 if total_chars > 0 else 0
print(f"Percentage of punctuation marks: {punctuation_percentage:.2f}%")
```


#### Print total number of lines

```python
total_lines = len(text_lines)
print(f"Total number of lines: {total_lines}")
```

#### How many words are in title case?

```python
title_case_count = 0
for word in all_words:
    cleaned_word = word.strip(string.punctuation)
    if cleaned_word.istitle():
        title_case_count += 1
print(f"Total words in title case: {title_case_count}")
```

#### 5. Remove numbers from the text

```python
cleaned_text = "".join(char for char in full_text if not char.isdigit())
print("\nCleaned text (Numbers removed) - Previewing first 500 characters:\n")
print(cleaned_text[:500] + "\n... [Truncated for preview] ...")
```
