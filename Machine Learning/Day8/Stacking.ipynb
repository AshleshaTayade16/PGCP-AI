```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier,DecisionTreeRegressor
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.metrics import accuracy_score,log_loss,r2_score,f1_score
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.linear_model import LogisticRegression,LinearRegression,ElasticNet
from sklearn.ensemble import StackingClassifier,RandomForestClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.compose import ColumnTransformer
from sklearn.compose import make_column_selector
from tqdm import tqdm
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeClassifier,plot_tree
```

```python
glass=pd.read_csv("D:\\AshleshaRuchika\\PGCP-AI\\Machine Learning\\Glass_Identification\\Glass.csv")
glass
```

```python
X,y=glass.drop("Type",axis=1),glass["Type"]
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=26,stratify=glass["Type"])
```

```python
lr=LogisticRegression()
dtc=DecisionTreeClassifier(random_state=26)
nb=GaussianNB()
rf=RandomForestClassifier(random_state=26)
stack=StackingClassifier(estimators=[('NB',nb),('LR',lr),('TREE',dtc)], final_estimator=rf)
stack.fit(X_train,y_train)
y_pred=stack.predict(X_test)
f1_score(y_test,y_pred,average="macro")
```

```python
svm=SVC()
stack=StackingClassifier(estimators=[('NB',nb),('SVM',svm),('TREE',dtc),('RF',rf)], final_estimator=svm,passthrough=True)
stack.fit(X_train,y_train)
y_pred=stack.predict(X_test)
f1_score(y_test,y_pred,average="macro")
```
