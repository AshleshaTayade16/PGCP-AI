```python
# alphas=np.linspace(0.001,10,30)
# params={'alpha':alphas}
# ridge=Ridge()
# gcv=GridSearchCV(estimator=ridge,param_grid=params,cv=kfold)
# gcv.fit(X, y)
# pd_cv=pd.DataFrame(gcv.cv_results_)
# pd_cv
alphas = np.linspace(0.001, 10, 30)
params = {'RIDGE__alpha': alphas} # Note the 'RIDGE__' prefix for the pipeline

# 2. Reuse your existing transformer (transf) from the previous step
ridge = Ridge()
pipe_ridge = Pipeline([("OHE", transf), ("RIDGE", ridge)])

# 3. Setup GridSearchCV with the pipeline
gcv = GridSearchCV(estimator=pipe_ridge, param_grid=params, cv=kfold, scoring='r2')
gcv.fit(X, y)

# 4. View results
pd_cv = pd.DataFrame(gcv.cv_results_)
pd_cv[['param_RIDGE__alpha', 'mean_test_score', 'rank_test_score']].sort_values('rank_test_score').head()
```
