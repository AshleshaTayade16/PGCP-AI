```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.manifold import TSNE
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer, make_column_selector
```

```python
train=pd.read_csv("D:\\AshleshaRuchika\\PGCP-AI\\Machine Learning\\Irrigation\\train.csv",index_col=0)
train
```

```python
test=pd.read_csv("D:\\AshleshaRuchika\\PGCP-AI\\Machine Learning\\Irrigation\\test.csv",index_col=0)
test
```

```python
y=train['Irrigation_Need']
X=train.drop('Irrigation_Need',axis=1)
```

```python
scaler=StandardScaler().set_output(transform='pandas')
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
ohe=OneHotEncoder(sparse_output=False,drop="first").set_output(transform="pandas")
trns=ColumnTransformer(transformers=[("OHE",ohe, make_column_selector
                                        (dtype_include=object))],remainder="passthrough",verbose_feature_names_out=False).set_output(transform="pandas")
X_trn_ohe=trns.fit_transform(X_train)
X_tst_ohe=trns.transform(X_test)
```

```python
tsne=TSNE(random_state=26,n_jobs=-1).set_output(transform='pandas')
X_tsne=tsne.fit_transform(X_trn_ohe)
tsne_data=pd.concat([X_tsne,y],axis=1)
```

```python
sns.scatterplot(data=tsne_data,x='tsne0',y='tsne1',hue='Irrigation_Need')
plt.show()
```
