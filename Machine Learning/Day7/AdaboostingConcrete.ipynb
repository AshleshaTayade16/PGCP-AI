```python
concrete=pd.read_csv("D:\\AshleshaRuchika\\PGCP-AI\\Machine Learning\\Concrete_Strength\\Concrete_Data.csv")
concrete
```

```python
X,y=concrete.drop("Strength",axis=1),concrete["Strength"]
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=26)
```

```python
le=LinearRegression()
el=ElasticNet(l1_ratio=1)
knn=KNeighborsRegressor()
dtr=DecisionTreeRegressor(random_state=26,max_depth=1)
for e in [dtr,le,el]:
    ada=AdaBoostRegressor(random_state=26,estimator=e)
    ada.fit(X_train,y_train)
    y_pred=ada.predict(X_test)
  
    print("for",e,":")
  
    print("r2_score=", r2_score(y_test,y_pred))
```
