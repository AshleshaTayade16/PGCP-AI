```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier,DecisionTreeRegressor
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.metrics import accuracy_score,log_loss,r2_score
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.linear_model import LogisticRegression,LinearRegression,ElasticNet
from sklearn.ensemble import BaggingRegressor, RandomForestClassifier, AdaBoostClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.compose import ColumnTransformer
from sklearn.compose import make_column_selector
from tqdm import tqdm
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeClassifier,plot_tree
```

```python
cancer=pd.read_csv("D:\\AshleshaRuchika\\PGCP-AI\\Machine Learning\\Wisconsin\\BreastCancer.csv")
cancer
```

```python
X,y=cancer.drop("Class",axis=1),cancer["Class"]
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=26,stratify=cancer["Class"])
```

```python
dtc=DecisionTreeClassifier(random_state=26,max_depth=1)
lr=LogisticRegression()
nb=GaussianNB()
ada=AdaBoostClassifier(random_state=26,estimator=dtc)
ada.fit(X_train,y_train)
y_pred=ada.predict(X_test)
y_pred_prob=ada.predict_proba(X_test)
print("ACC=", accuracy_score(y_test,y_pred))
print("log loss=", log_loss(y_test,y_pred_prob))



```
