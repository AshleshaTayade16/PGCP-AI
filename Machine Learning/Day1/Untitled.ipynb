```python
import numpy as np
import os
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score
from sklearn.preprocessing import PolynomialFeatures
from sklearn.impute import SimpleImputer
os.chdir("D:\\AshleshaRuchika\\PGCP-AI\\Machine Learning\\Datasets")
```

```python
pizza=pd.read_csv("pizza.csv")
pizza
```

```python
X=pizza[["Promote"]]
y=pizza["Sales"]
lr=LinearRegression()
lr.fit(X,y)
```

```python
lr.coef_,lr.intercept_
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=26)
lr=LinearRegression()
lr.fit(X_train,y_train)
y_pred=lr.predict(X_test)
print("MAE=",mean_absolute_error(y_test,y_pred))
print("MSE=",root_mean_squared_error(y_test,y_pred))
print("R2=",r2_score(y_test,y_pred))
```

```python
boston=pd.read_csv("Boston.csv")
X=boston.drop("medv",axis=1)
y=boston["medv"]
poly=PolynomialFeatures(degree=2,include_bias=False).set_output(transform='pandas')
X_poly=poly.fit_transform(X)
X_train,X_test,y_train,y_test=train_test_split(X_poly,y,test_size=0.3,random_state=26)
lr=LinearRegression()
lr.fit(X_train,y_train)
y_pred=lr.predict(X_test)
print("R2=",r2_score(y_test,y_pred))
```

```python
housing=pd.read_csv("Housing.csv")
housing
```

```python
dum_hous=pd.get_dummies(housing,dtype=int,drop_first=True)
print(dum_hous)
X,y=dum_hous.drop("price",axis=1),dum_hous["price"]
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=26)
lr=LinearRegression()
lr.fit(X_train,y_train)
y_pred=lr.predict(X_test)
print("R2=",r2_score(y_test,y_pred))
```

```python
X,y=chem.drop("Yield",axis=1),chem["Yield"]
mean_imp=SimpleImputer(strategy="mean")
med_imp=SimpleImputer(strategy="median")
X_imp=mean_imp.fit_transform(X)
X_train,X_test,y_train,y_test=train_test_split(X_imp,y,test_size=0.3,random_state=26)
lr=LinearRegression()
lr.fit(X_train,y_train)
y_pred=lr.predict(X_test)
# print("MAE=",mean_absolute_error(y_test,y_pred))
# print("MSE=",root_mean_squared_error(y_test,y_pred))
print("R2=",r2_score(y_test,y_pred))
```

```python
mean_imp=SimpleImputer(strategy="mean").set_output(transform="pandas")
X_trn_imp=mean_imp.fit_transform(X_train)
X_tst_imp=mean_imp.transform(X_test)
lr.fit(X_trn_imp,y_train)
y_pred=lr.predict(X_tst_imp)
print("R2=",r2_score(y_test,y_pred))
```

```python
med_imp=SimpleImputer(strategy="median").set_output(transform="pandas")
X_trn_imp=med_imp.fit_transform(X_train)
X_tst_imp=med_imp.transform(X_test)
lr.fit(X_trn_imp,y_train)
y_pred=lr.predict(X_tst_imp)
print("R2=",r2_score(y_test,y_pred))
```
