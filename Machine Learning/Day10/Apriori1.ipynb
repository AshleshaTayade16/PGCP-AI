```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
from ipywidgets import interact, widgets

# 1. Load and process data correctly
with open("groceries.csv", "r") as f:
    groceries_content = f.read()

# Split by line first, then split by comma
groceries_list = [line.split(",") for line in groceries_content.strip().split("\n")]

# 2. Encode the data
te = TransactionEncoder()
te_ary = te.fit(groceries_list).transform(groceries_list)
fp_df = pd.DataFrame(te_ary, columns=te.columns_)

# 3. Define the interactive function
def gen_rules(permissible_support, confidence_threshold):
    # CRITICAL: Use the encoded DataFrame 'fp_df' here, NOT the list
    itemsets = apriori(fp_df, min_support=permissible_support, use_colnames=True)
    
    if itemsets.empty:
        return "No frequent itemsets found."
        
    rules = association_rules(itemsets, metric='confidence', min_threshold=confidence_threshold)
    
    if rules.empty:
        return "No association rules found."
        
    rules = rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
    relv_rules = rules[rules['lift'] > 1]
    
    # CRITICAL: You MUST return the result to display it in the widget
    return relv_rules.sort_values('lift', ascending=False)

# 4. Call interact
interact(gen_rules, 
         permissible_support=widgets.FloatSlider(value=0.2, min=0.01, max=0.5, step=0.01, description="Support:"),
         confidence_threshold=widgets.FloatSlider(value=0.6, min=0.1, max=1.0, step=0.05, description="Confidence:"))
```
