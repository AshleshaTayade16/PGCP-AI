```python
import numpy as np
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, ElasticNet
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score
os.chdir("D:\\AshleshaRuchika\\PGCP-AI\\Machine Learning\\Concrete_Strength")
```

```python
concretedata=pd.read_csv("Concrete_Data.csv")
concretedata
```

```python
#infercing
# 1. fit the best model on the whole data
X=concretedata.drop("Strength",axis=1)
y=concretedata["Strength"]
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=26)
lr=LinearRegression()
lr.fit(X_train,y_train)
y_pred=lr.predict(X_test)
print("MAE=",mean_absolute_error(y_test,y_pred))
print("MSE=",root_mean_squared_error(y_test,y_pred))
print("R2=",r2_score(y_test,y_pred))
```

```python
bm=ElasticNet(alpha=5,l1_ratio=0.26158)
bm.fit(X,y)
```
