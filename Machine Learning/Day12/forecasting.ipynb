```python
df=pd.read_csv("D:\\AshleshaRuchika\\PGCP-AI\\Machine Learning\\Datasets\\monthly-milk-production-pounds-p.csv")
y_train=df['Milk'].iloc[:-12]
y_test=df['Milk'].iloc[-12:]
y_train.shape,y_test.shape
```

```python
def MA(span):
    fcast=y_train.rolling(span).mean()
    MA=fcast.iloc[-1]
    y_pred=pd.Series(MA.repeat(len(y_test)),index=y_test.index)
    plt.plot(y_pred,c='blue',label='forecast')
    plt.plot(y_test,c='green',label='Test')
    rmse=np.round(root_mean_squared_error(y_test,y_pred),2)
    plt.text(164,900,f'RMSE={rmse}')
    plt.legend(loc='best',frameon=True)
    plt.show()
interact(MA,span=(2,7))
```
