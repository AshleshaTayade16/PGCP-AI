```python
glass=pd.read_csv("D:\\AshleshaRuchika\\PGCP-AI\\Machine Learning\\Cases (1)\\Glass_Identification\\Glass.csv")
glass.head(5)
```

```python
df=glass.drop("Type",axis=1)
```

```python
clf=IsolationForest(contamination=0.05,random_state=26)
clf.fit(df)
predictions=clf.predict(df)
predictions
```

```python
df['outlier_status'] = predictions
outliers = df[df['outlier_status'] == -1]
outliers
```

```python
count_table = pd.crosstab(glass['Type'], df['outlier_status'],margins=True)
count_table
```

```python
print("percentage of outliers="+str((predictions<0).mean()*100)+"%")
```

```python
from xgboost import XGBClassifier
from sklearn.model_selection import StratifiedKFold,GridSearchCV
```

```python
in_glass=df[predictions!=-1]
le=LabelEncoder()
kfold=StratifiedKFold(n_splits=5,shuffle=True,random_state=26)
in_glass['Type']=le.fit_transform(in_glass['Type'])
X,y=in_glass.drop('Type',axis=1),in_glass['Type']
params={'learning_rate':np.linspace(0.0001,0.9,30),'n_estimators':[10,25,50,75,100],'max_depth':[2,3,4,5,6]}
gbm=XGBClassifier(random_state=26,n_jobs=-1)
gcv=GridSearchCV(gbm,param_grid=params,cv=kfold,scoring='neg_log_loss')

gcv.fit(X,y)
gcv.best_params_,gcv.best_score_
```
