```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score,log_loss, classification_report,roc_auc_score
from sklearn.compose import make_column_selector, ColumnTransformer
from sklearn.naive_bayes import GaussianNB, BernoulliNB
from tqdm import tqdm
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis, LinearDiscriminantAnalysis
```

```python
flower=pd.read_csv("D:\\AshleshaRuchika\\PGCP-AI\\Machine Learning\\Datasets\\iris.csv")
flower
```

```python
X,y=flower.drop('Species',axis=1), flower['Species']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=26,stratify=flower['Species'])
```

```python
lda=LinearDiscriminantAnalysis()
lda.fit(X_train,y_train)# cxalculate all the apriori probabilities
y_pred_prob=lda.predict_proba(X_test)# calculates posterior probabilities
print("log_loss=",log_loss(y_test,y_pred_prob))
y_pred=lda.predict(X_test)
print("Accuracy=",accuracy_score(y_test,y_pred))
```

```python
qda=QuadraticDiscriminantAnalysis(reg_param=0.9)
qda.fit(X_train,y_train)# cxalculate all the apriori probabilities
y_pred_prob=qda.predict_proba(X_test)# calculates posterior probabilities
print("log_loss=",log_loss(y_test,y_pred_prob))
y_pred=qda.predict(X_test)
print("Accuracy=",accuracy_score(y_test,y_pred))
```

```python
score=[]
for p in np.linspace(0.0001,1,100):
    qda=QuadraticDiscriminantAnalysis(reg_param=p)
    qda.fit(X_train,y_train)
    y_pred_prob=qda.predict_proba(X_test)
    score.append([p,log_loss(y_test,y_pred_prob)])
df_scores=pd.DataFrame(score,columns=['ref-param','score'])
df_scores.sort_values('score')
```
