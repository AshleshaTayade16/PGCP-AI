```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import RidgeCV, LassoCV, ElasticNetCV
from sklearn.model_selection import train_test_split, GridSearchCV, KFold, cross_val_score
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer, make_column_selector
from sklearn.decomposition import PCA
from sklearn.ensemble import StackingRegressor
from sklearn.impute import SimpleImputer
from sklearn.metrics import accuracy_score, r2_score, f1_score
from sklearn.pipeline import Pipeline
import os
os.chdir(r"D:\\AshleshaRuchika\\PGCP-AI\\Machine Learning\\Datasets")
```

```python
df = pd.read_csv("Boston.csv")
df
```

```python
X,y = df.drop('medv', axis=1),df['medv']
X.shape, y.shape
```

```python
kfold = KFold(shuffle=True, n_splits=5, random_state=26)
```

```python
scalar = StandardScaler()
```

```python
rdg = RidgeCV(alphas=np.logspace(-3,3,100), cv=5)
lso = LassoCV(alphas=np.logspace(-3,3,100), cv=5)
ent = ElasticNetCV(cv=5, random_state=26, l1_ratio=np.linspace(0.001,1,10), alphas=np.logspace(-3,3,100))
```

```python
pipe_rdg = Pipeline(steps=[('scalar',scalar),('rdg',rdg)])
pipe_lso = Pipeline(steps=[('scalar',scalar),('lso',lso)])
pipe_ent = Pipeline(steps=[('scalar',scalar),('ent',ent)])
```

```python
score_rdg = cross_val_score(pipe_rdg, X, y, cv=kfold)
score_lso = cross_val_score(pipe_lso, X, y, cv=kfold)
score_ent = cross_val_score(pipe_ent, X, y, cv=kfold)
```

```python
score_rdg.mean(), score_lso.mean(), score_ent.mean()
```
