```python
import numpy as np
import pandas as pd
import os
from sklearn.linear_model import LogisticRegression
from sklearn.linear_model import LogisticRegressionCV
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix,accuracy_score,f1_score,recall_score,precision_score,classification_report
os.chdir("D:\\AshleshaRuchika\\PGCP-AI\\Machine Learning\\Sonar")
```

```python
sonar=pd.read_csv("Sonar.csv")
X,y=sonar.drop('Class', axis=1), sonar['Class'] 
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=26,stratify=sonar['Class'])
lr=LogisticRegression()
lr.fit(X_train,y_train)
y_pred_prob=lr.predict_proba(X_test)
print("Shape: ",y_pred_prob.shape)
y_pred_prob
```

```python
confusion_matrix(y_test,y_pred)
print(recall_score(y_test,y_pred,pos_label="M"))
print(recall_score(y_test,y_pred,pos_label="R"))
```

```python
print(f1_score(y_test,y_pred,pos_label="M"))
print(f1_score(y_test,y_pred,pos_label="R"))
```

```python
solvers=["lbfgs","liblinear","newton-cg","newton-choesky","sag","saga"]
# Cs.np.linspace(0.001,5,20)
scores=[]
for s in solvers:
    lr=LogisticRegression()
    lr.fit(X_train,y_train)
    y_pred=lr.predict(X_test)
    scores.append([s,f1_score(y_test,y_pred,average="macro")])
df_scores=pd.DataFrame(scores,columns=["solver","score"])
df_scores.sort_values("score",ascending=False)
```

```python
solvers=["lbfgs","liblinear","newton-cg","newton-cholesky","sag","saga"]
Cs=np.linspace(0.001,5,20)
scores=[]
for s in solvers:
    for c in Cs:
        lr=LogisticRegression(solver=s,C=c,max_iter=500)
        lr.fit(X_train,y_train)
        y_pred=lr.predict(X_test)
        scores.append([s,c,f1_score(y_test,y_pred,average="macro")])
df_scores=pd.DataFrame(scores,columns=["solver","C","score",]) 
df_scores.sort_values("score",ascending=False)
```
