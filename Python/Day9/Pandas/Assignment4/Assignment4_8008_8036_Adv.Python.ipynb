```python
# Q2. Read csv file solar.csv
# a.	Display all the plants with capacity > 500 MW
# b.	Display plant details for New York plant
# c.	For all the plants display Average MW Per Plant and Generation (GWh)
# d.	Sort the plants by Generation (GWh) (ascending) and Installed Capacity (MW)(descending)
# e.	List top 5 plants for their Generation
# f.	Display all plant details for states – California, Nevada, Arizona and Texas
# g.	Group the plants by region, find – min/max capacity plant for each group
# h.	Add following information to the existing details for the plants
# 	North Carolina	148	669	4.5	1162	North
```

```python
df1=pd.read_csv("solar.csv")
df1
```

```python
#a.Display all the plants with capacity > 500 MW
filt=df1["Installed Capacity (MW)"]>500
df1.loc[filt,["State","Number of Solar Plants","Installed Capacity (MW)","Average MW Per Plant","Generation (GWh)","Region"]]
```

```python
#b.Display plant details for New York plant
filt=df1["State"]=="New York"
df1.loc[filt,["State","Number of Solar Plants","Installed Capacity (MW)","Average MW Per Plant","Generation (GWh)","Region"]]
```

```python
#c.For all the plants display Average MW Per Plant and Generation (GWh)
df1.loc[:,["Average MW Per Plant","Generation (GWh)"]]
```

```python
#d.Sort the plants by Generation (GWh) (ascending) and Installed Capacity (MW)(descending)
df1.sort_values(by=["Generation (GWh)", "Installed Capacity (MW)"],ascending=[True,False])
```

```python
#e.List top 5 plants for their Generation
df1.head()
```

```python
# f.Display all plant details for states – California, Nevada, Arizona and Texas
filt=df1["State"].isin(["California","Nevada","Arizona","Texas"])
df1.loc[filt,["State","Number of Solar Plants","Installed Capacity (MW)","Average MW Per Plant","Generation (GWh)","Region"]]
```

```python
#g.Group the plants by region, find – min/max capacity plant for each group
df1_groups =df1.groupby(df1['Region'])
df1_groups["Installed Capacity (MW)"].agg(["min","max"])
```

```python
# h.Add following information to the existing details for the plants
# North Carolina	148	669	4.5	1162	North
row_data = {"State":"North Carolina","Number of Solar Plants":148,"Installed Capacity (MW)":669,"Average MW Per Plant":4.5,"Generation (GWh)":11.62,
            "Region":"North"}
df1=pd.concat([df1, pd.DataFrame([row_data])], ignore_index=True)
df1
```

```python
# Q.3Read csv gapminder2007.csv
# a.	Display data for all Asian countries
# b.	Display top and bottom 15 rows
# c.	Show all the rows where life expectancy is more than 50 years and less than 80 years
# d.	Show data for India, America and France
# e.	Show data for all countries where name starts with ‘A’
# f.	Show min/max and mean GDP and Population for each continent
# g.	Show population and life expectancy for all countries 
# h.	Sort the data by country names (ascending) and GDP (descending)
# i.	Find top 20 most populous countries
# j.	Delete all rows for a specific country
# k.	Change all column names to title case
# l.	Change column names: pop to Population, gdpPercap to GDP_Percap
# m.	Increase Life Expectancy by 2 years for all African countries
# n.	Find all the rows that contain NA values
# o.	Fill NA values with 0 for numeric columns
```

```python
df2=pd.read_csv("gapminder2007.csv")
df2
```

```python
# a.Display data for all Asian countries
filt=df2["continent"]=="Asia"
df2.loc[filt,:]
```

```python
# b.Display top and bottom 15 rows
df2.head(15)
```

```python
df2.tail(15)
```

```python
# c.Show all the rows where life expectancy is more than 50 years and less than 80 years
filt=(df2["lifeExp"]>50) & (df2["lifeExp"]<80)
df2.loc[filt,["country","pop","continent","lifeExp","gdpPercap"]]
```

```python
# d.Show data for India, America and France
filt=df2["country"].isin(["India","America","France"])
df2.loc[filt,["country","pop","continent","lifeExp","gdpPercap"]]
```

```python
# e.Show data for all countries where name starts with ‘A’
filt=df2["country"].str.startswith("A")
df2.loc[filt,:]
```

```python
# f.Show min/max and mean GDP and Population for each continent
df2_groups =df2.groupby(df2['continent'])
df2_groups[["pop","gdpPercap"]].agg(["min","max","mean"])
```

```python
# g.Show population and life expectancy for all countries 
filt=df2["country"]
df2.loc[:,["country","pop","lifeExp"]]
```

```python
# h.Sort the data by country names (ascending) and GDP (descending)
df2.sort_values(by=["country", "gdpPercap"],ascending=[True,False])
```

```python
# i.Find top 20 most populous countries
df2.nlargest(20,"pop")
```

```python
# j.Delete all rows for a specific country
df2 =df2[df2['country']!='Afghanistan']
df2
```

```python
# k.Change all column names to title case
df2.columns=[col.title() for col in df2.columns]
df2.columns
```

```python
# l.Change column names: pop to Population, gdpPercap to GDP_Percap
df2=df2.rename(columns={"pop":"Population","gdpPercap":"GDP_Percap"})
df2
```

```python
# m.Increase Life Expectancy by 2 years for all African countries
df2.loc[df2["Continent"]=="Africa","Lifeexp"]+=2
df2
```

```python
# n.Find all the rows that contain NA values
rows_with_nan=df2[df2.isnull().any(axis=1)]
rows_with_nan
```

```python
# o.Fill NA values with 0 for numeric columns
numeric_cols=df2.select_dtypes(include=["number"]).columns
df2[numeric_cols]=df2[numeric_cols].fillna(0)
df2
```
