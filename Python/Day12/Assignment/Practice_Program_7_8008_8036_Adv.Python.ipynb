```python
import plotly.express as px
import seaborn as sns
```

```python
dir(px.data)
```

```python
# 1.Draw a bar chart showing GDP percentage for European countires. [use gdpminder dataset]
# 	Use different semantics as orientation, bar mode, bar norm, text to improve your chart
# 	For the same dataset show GDP over the years using animation frame as Country 
df = px.data.gapminder()
df
```

```python
filt= df.loc[df["continent"]== "Europe"]
filt
```

```python
fig=px.bar(filt,x=filt['country'],y=filt['gdpPercap'],color=filt['year'])
fig.show()
```

```python
fig=px.bar(filt,x=filt['country'],y=filt['gdpPercap'],animation_frame=filt['year'],text_auto='.2s')
fig.update_traces(textfont_size=12, textangle=0, textposition="outside", cliponaxis=False)
fig.show()
```

```python
fig = px.line(df, x=df['year'], y=df["gdpPercap"], animation_frame=df["country"], )
fig.update_layout(xaxis_title="Year", yaxis_title="GDP per Capita")
fig.show()
```

```python
fig=px.bar(filt, x='country', y='gdpPercap', text='gdpPercap', animation_frame='year', color='country')
fig.update_layout(title='gdp per country',showlegend = True,hovermode='y unified')
fig.show()
```

```python
# 2.Draw a Histogram showing votes secured by 3 candidates - Coderre, Joly, Bergeron
# 	Use semantics -bar mode, color, facet_row to improve your chart. Create a cumulative graph as well. [use election dataset]
df=px.data.election()
df
```

```python
fig=px.histogram(df,x=['Coderre', 'Joly','Bergeron'],color='winner',barmode='overlay',facet_row='winner')
fig.show()
```

```python
# 3.Draw a scatter plot to show relationship between iris petal length and sepal width
df = px.data.iris()
df
```

```python
fig=px.scatter(df, x='petal_length', y ='sepal_width', color='species', symbol='species')
```
