```python
import torch
import torch.nn as nn

# 1. Define the Input (Let's say we have 3 features, like our running example)
# Creating a tensor of size (1 row, 3 columns)
X = torch.tensor([[1.0, -2.0, 3.0]]) 

# 2. Define a Single Layer Neural Network (Linear Layer)
# in_features=3 (our inputs), out_features=1 (our single neuron)
single_layer = nn.Linear(in_features=3, out_features=1)

# PyTorch randomly initializes the Weights and Biases for us!
print("Initial Weights:", single_layer.weight.data)
print("Initial Bias:", single_layer.bias.data)

# 3. Perform the Forward Pass (Calculate z = W*X + b)
z = single_layer(X)
print("\nRaw Output (z):", z.item())

# 4. Apply Activation Functions
# Sigmoid
sigmoid_activation = torch.sigmoid(z)
print("After Sigmoid (between 0 and 1):", sigmoid_activation.item())

# Tanh
tanh_activation = torch.tanh(z)
print("After Tanh (between -1 and 1):", tanh_activation.item())

# ReLU
relu_activation = torch.relu(z)
print("After ReLU (0 if negative, unchanged if positive):", relu_activation.item())
```
