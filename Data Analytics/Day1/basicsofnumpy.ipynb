```python
#Basics Usage
a=np.arange(0,10,2)
b= np.linspace(0,10,5)
print("arange(0,10,2)=",a)
print("linspace(0,10,5)=",b)
```

```python
#float behvaiour
print("\narange with float step(may miss endpoint):")
print(np.arange(0,1,0.2))
print("\narange with exact number of points:")
print(np.linspace(0,1,6))
```

```python
#endpoint control
print("\nlinspace without endpoint:")
print(np.linspace(0,10,5,endpoint=False))
```

```python
#length comparison
print("\nNumber of elements:")
print("len(arange):",len(np.arange(0,10,2)))
print("len(linspace):",len(np.linspace(0,10,5)))
```

```python
#precision demonstration
print("\nPrecision check:")
arr =np.arange(0,1,0.1)
lin= np.linspace(0,1,10)
print("arange:",arr)
print("linspace:",lin)
```

```python
#filterating
arr2=np.arange(55,65)
print("original Array:\n",arr2)
```

```python
#apply condition
mask= arr2>58
print("\nCondition(arr>58):\n",mask)
```

```python
#boolean mask
print("\nBoolean Mask(true values will be selected):\n",mask)
```

```python
#filtered values
filtered= arr2[mask]
print("\nFiltered Values(>58):\n",filtered)
```

```python
#Direct filtering 
print("\nDirect filtering :\n",arr2[arr2>58])
```

```python
#multiple condition
print("\nValues between 57 and 62\n", arr2[(arr2>57) & (arr2<63)])
```

```python
#shallo copy vr deep copy
a= np.array([50, 60,70,90])
a
```

```python
#shallow copy
b=a[0:5]
print("Original a:",a)
print("\nShallow copy b:",b)
```

```python
b[0]= 667
print("\nAfter modifying b:")
print("a:",a)
print("b:",b)
```

```python
#deep copy
c=a.copy()
c[1]=657
print("\nnAfter modifying c:")
print("a:",a)
print("c:",c)
```
