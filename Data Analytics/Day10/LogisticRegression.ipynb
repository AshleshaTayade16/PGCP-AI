```python
#Logistic REgression: Intuition
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Given model prameters
b0=-10
b1=0.2

#Dog weights (our sample data)
weights=np.array([20,30,40,50,60,70])

#Logit values
z=b0+b1*weights

#Sigmoid function
def sigmoid(z):
    return 1/(1+np.exp(-z))

#Probabilities
probs=sigmoid(z)   #If P>=0.5 output class=1  else class=0
# probs

#Print table
print("Weight     Logit(z)     Probability     Result")
for w, logit,p in zip(weights, z, probs):
    result="Obese" if p>=0.5 else "Not Obese"
    print(f"w{w:6}   {logit:8.2f}    {p:10.4f}       {result}")


#Create smooth curve
x_vals=np.linspace(0,80,200)
z_vals=b0+b1*x_vals
y_vals=sigmoid(z_vals)

#Plot

plt.figure()
plt.plot(x_vals,y_vals)    #sigmoid curve
plt.scatter(weights,probs)   #actual pointts
#Decision boundary line
plt.axhline(y=0.5)
plt.xlabel("Dog weight(kg)")
plt.ylabel("Probability of Obesity")
plt.title("Logistic Regression (Dog Obesity)")
plt.show()

```
