```python
# 24. Write a Pandas program to generate time series combining day and intraday
from pandas.tseries.offsets import Day, Hour, Minute
start_time = pd.Timestamp('2026-04-11 08:00:00')
combined_offset = Day(1) + Hour(3)
time_series = [start_time + i * combined_offset for i in range(5)]
dt_index = pd.DatetimeIndex(time_series)
print("Time Series with Combined Offsets (1 Day + 3 Hours):")
print(dt_index)
alt_range = pd.date_range(start=start_time, periods=5, freq='1D3H')
print("\nAlternative using String Aliases ('1D3H'):")
print(alt_range)
```

```python
# 27. Write a Pandas program to calculate one, two, three business day(s) from a specified date. Also find the next business month end from a specific date.
from pandas.tseries.offsets import BDay, BMonthEnd
start_date = pd.Timestamp('2026-04-11')
bus_day_1 = start_date + BDay(1)
bus_day_2 = start_date + BDay(2)
bus_day_3 = start_date + BDay(3)
next_bmonth_end = start_date + BMonthEnd(1)
print(f"Starting Date: {start_date.date()} ({start_date.day_name()})")
print("-" * 30)
print(f"1 Business Day later:  {bus_day_1.date()} ({bus_day_1.day_name()})")
print(f"2 Business Days later: {bus_day_2.date()} ({bus_day_2.day_name()})")
print(f"3 Business Days later: {bus_day_3.date()} ({bus_day_3.day_name()})")
print("-" * 30)
print(f"Next Business Month End: {next_bmonth_end.date()} ({next_bmonth_end.day_name()})")
```

```python
# 28. Write a Pandas program to create a period index represent all monthly
# boundaries of a given year. Also print start and end time for each period object in
# the said index.
year = 2026
monthly_periods = pd.period_range(start=f'{year}-01', end=f'{year}-12', freq='M')
period_data = []
for period in monthly_periods:
    period_data.append({
        'Period': period,
        'Start Time': period.start_time,
        'End Time': period.end_time
    })
df = pd.DataFrame(period_data)
print(f"Monthly Boundaries for {year}:")
print(df.to_string(index=False))
```
