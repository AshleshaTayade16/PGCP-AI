```python
# 16. Write a Pandas program to get the items of a given series not present in another given series.

sr1=pd.Series([1,2,3,4,5])
sr2=pd.Series([2,4,6,8,10])
print("Series 1: \n",sr1)
print("\nSeries 2: \n",sr2)

new_series=sr1[~sr1.isin(sr2)]
print("\nItems of sr1 not present in sr2: \n",new_series)
```

```python
# 17. Write a Pandas program to get the items which are not common of two given series.

sr1=pd.Series([1,2,3,4,5])
sr2=pd.Series([2,4,6,8,10])
print("Series 1: \n",sr1)
print("\nSeries 2: \n",sr2)

sr_union = pd.Series(pd.concat([sr1, sr2]))
sr_intersection = pd.Series(list(set(sr1) & set(sr2)))

new_series=sr_union[~sr_union.isin(sr_intersection)]
print("\nItems of a given series not present in another given series:\n", new_series)
```

```python
# 33. Write a Pandas program to replace missing white spaces in a given string with the
# least frequent character.
str1 = 'abc def abcdef icd'
ser = pd.Series(list(str1))
freq = ser.value_counts()

least_freq_char = freq.drop(' ').idxmin()
result = str1.replace(' ', least_freq_char)

print("Original string:")
print(str1)
print("\nLeast frequent character:", least_freq_char)
print("\nResult:")
print(result)
```

```python
# 34. Write a Pandas program to compute the autocorrelations of a given numeric series.
# From Wikipedia:
# Autocorrelation, also known as serial correlation, is the correlation of a signal with a
# delayed copy of itself as a function of delay. Informally, it is the similarity between
# observations as a function of the time lag between them.
np.random.seed(42)
s = pd.Series(np.random.normal(10, 5, 15))

print("Original series:")
print(s)

autocorr_values = [round(s.autocorr(lag=i), 2) for i in range(1, 11)]

print("\nAutocorrelations of the said series:")
print(autocorr_values)
```
