```python
#linear programming problem(LPP)-chairs/tables
from pulp import *
```

```python
#create the problem varible to cinatin the probelm data
model=LpProblem("FactoryProductionProblem",LpMaximize)
```

```python
#create varibles for Chairs and Tables
#parameters:Name,lower limit,upper limit,data type
chairs=LpVariable("Chairs",0,None,LpInteger)
tables=LpVariable("Tables",0,None,LpInteger)
```

```python
#create maximize objective function(profit)
model+=300*chairs +1000*tables,"Profit"
```

```python
#create constaints
model+=5*chairs+10*tables<=400,"Wood Constraint"
model+=2*chairs+6*tables<=300,"Time Constraint"
model.solve()
```

```python
#each of the varibels is printed with its resolved optimum value 
for v in model.variables():
    print(v.name,"=",v.varValue)
print("Total Profit:",value(model.objective))
```
