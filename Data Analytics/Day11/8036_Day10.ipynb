```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = {
    'Date': pd.date_range(start='2023-01-01', periods=100),
    'Open': np.random.uniform(100, 110, 100),
    'Close': np.random.uniform(100, 110, 100),
    'High': np.random.uniform(111, 120, 100),
    'Low': np.random.uniform(90, 99, 100),
    'Volume': np.random.randint(1000000, 5000000, 100),
    'Adj Close': np.random.uniform(100, 110, 100)
}
df = pd.DataFrame(data).set_index('Date')
start_date = '2023-01-01'
end_date = '2023-02-15'
```

```python
# 1. Write a Pandas program to create a line plot of the historical stock prices of
# Alphabet Inc. between two specific dates.
line_plot_data = df.loc[start_date:end_date, 'Close']
plt.figure(figsize=(8, 6))
line_plot_data.plot(color='green')
plt.title(f'Alphabet Inc. Historical Stock Price\n({start_date} to {end_date})')
plt.xlabel('Date')
plt.ylabel('Closing Price (USD)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# 2. Write a Pandas program to create a line plot of the opening, closing stock
# prices of Alphabet Inc. between two specific dates.
compare_prices = df.loc[start_date:end_date, ['Open', 'Close']]
compare_prices.plot(figsize=(8, 6))
plt.title(f'Alphabet Inc. Open vs. Close Prices\n({start_date} to {end_date})')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(['Opening Price', 'Closing Price'])
plt.tight_layout()
plt.show()
```

```python
# 3. Write a Pandas program to create a bar plot of the trading volume of Alphabet
# Inc. stock between two specific dates.
volume_subset = df.loc[start_date:end_date, 'Volume']
volume_subset.index = volume_subset.index.strftime('%Y-%m-%d')
volume_subset.plot(kind='bar', figsize=(12, 6), color='teal')
plt.title(f'Alphabet Inc. Trading Volume: {start_date} to {end_date}')
plt.xlabel('Date')
plt.ylabel('Volume')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 4. Write a Pandas program to create a bar plot of opening, closing stock prices of
# Alphabet Inc. between two specific dates.
prices_subset = df.loc[start_date:end_date, ['Open', 'Close']]
prices_subset.index = prices_subset.index.strftime('%Y-%m-%d')
prices_subset.plot(kind='bar', figsize=(12, 7), color=['#1f77b4', '#ff7f0e'])
plt.title(f'Alphabet Inc. Open vs. Close Prices ({start_date} to {end_date})')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(["Opening Price", "Closing Price"])
plt.tight_layout()
plt.show()
```

```python
# 5. Write a Pandas program to create a stacked bar plot of opening, closing stock
# prices of Alphabet Inc. between two specific dates.
stacked_data = df.loc[start_date:end_date, ['Open', 'Close']]
stacked_data.index = stacked_data.index.strftime('%Y-%m-%d')
stacked_data.plot(kind='bar', stacked=True, figsize=(12, 6), color=['#4285F4', '#FBBC05'])
plt.title(f'Stacked Open and Close Prices of Alphabet Inc.\n({start_date} to {end_date})')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend(["Open", "Close"], loc='upper left')
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# 6. Write a Pandas program to create a horizontal stacked bar plot of opening,
# closing stock prices of Alphabet Inc. between two specific dates.
horizontal_data = df.loc[start_date:end_date, ['Open', 'Close']]
horizontal_data.index = horizontal_data.index.strftime('%Y-%m-%d')
horizontal_data.plot(kind='barh', stacked=True, figsize=(8, 10), color=['#34A853', '#EA4335'])
plt.title(f'Horizontal Stacked Open/Close Prices: {start_date} to {end_date}')
plt.xlabel('Price (USD)')
plt.ylabel('Date')
plt.legend(["Open", "Close"], loc='lower right')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# 7. Write a Pandas program to create a histograms plot of opening, closing, high,
# low stock prices of Alphabet Inc. between two specific dates.
df.loc[start_date:end_date, ['Open', 'Close', 'High', 'Low']].plot(kind='hist', alpha=0.5, bins=15, title='Price Histograms')
plt.xlabel('Price (USD)')
plt.show()
```

```python
# 8. Write a Pandas program to create a stacked histograms plot of opening,
# closing, high, low stock prices of Alphabet Inc. between two specific dates.
hist_data = df.loc[start_date:end_date, ['Open', 'Close', 'High', 'Low']]
hist_data.plot(kind='hist', stacked=True, bins=20, figsize=(10, 6), edgecolor='white')
plt.title(f'Stacked Histogram of Alphabet Inc. Prices ({start_date} to {end_date})')
plt.xlabel('Price (USD)')
plt.ylabel('Frequency')
plt.legend(['Open', 'Close', 'High', 'Low'])
plt.show()
```

```python
# 9. Write a Pandas program to draw a horizontal and cumulative histograms plot
# of opening stock prices of Alphabet Inc. between two specific dates.
open_prices = df.loc[start_date:end_date, 'Open']
open_prices.plot(kind='hist', 
                 cumulative=True, 
                 orientation='horizontal', 
                 figsize=(10, 6), 
                 color='skyblue', 
                 edgecolor='white')
plt.title(f'Cumulative Horizontal Histogram: Alphabet Inc. Opening Prices\n({start_date} to {end_date})')
plt.xlabel('Frequency (Cumulative Count of Days)')
plt.ylabel('Opening Price (USD)')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

```python
# 10. Write a Pandas program to create a stacked histograms plot of opening,
# closing, high, low stock prices of Alphabet Inc. between two specific dates with
# more bins.
price_data = df.loc[start_date:end_date, ['Open', 'Close', 'High', 'Low']]
price_data.plot(kind='hist', 
                stacked=True, 
                bins=50, 
                figsize=(8, 6), 
                alpha=0.8, 
                edgecolor='black',
                linewidth=0.5)

plt.title(f'Alphabet Inc. Price Distribution (Stacked, 50 Bins)\n{start_date} to {end_date}')
plt.xlabel('Price (USD)')
plt.ylabel('Frequency')
plt.legend(['Open', 'Close', 'High', 'Low'])
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# 11. Write a Pandas program to create a stacked histograms plot with more bins
# of opening, closing, high, low stock prices of Alphabet Inc. between two specific
# dates.
# Functionally similar to 10, typically implies higher granularity
detailed_hist_data = df.loc[start_date:end_date, ['Open', 'Close', 'High', 'Low']]
detailed_hist_data.plot(kind='hist', 
                         stacked=True, 
                         bins=100, 
                         figsize=(8, 6), 
                         edgecolor='none') 
plt.title(f'Alphabet Inc. High-Granularity Price Distribution (100 Bins)\n{start_date} to {end_date}')
plt.xlabel('Price (USD)')
plt.ylabel('Frequency (Count)')
plt.legend(['Open', 'Close', 'High', 'Low'])
plt.tight_layout()
plt.show()
```

```python
# 12. Write a Pandas program to create a plot of stock price and trading volume of
# Alphabet Inc. between two specific dates.
price_volume_data = df.loc[start_date:end_date, ['Close', 'Volume']]
price_volume_data.plot(subplots=True, 
                       figsize=(8, 6), 
                       title=f'Alphabet Inc. Price and Volume\n({start_date} to {end_date})',
                       color=['#4285F4', '#34A853']) 
plt.xlabel('Date')
plt.tight_layout()
plt.show()
```

```python
# 13. Write a Pandas program to create a plot of Open, High, Low, Close, Adjusted
# Closing prices and Volume of Alphabet Inc. between two specific dates.
columns_to_plot = ['Open', 'High', 'Low', 'Close', 'Adj Close', 'Volume']
all_metrics = df.loc[start_date:end_date, columns_to_plot]
all_metrics.plot(subplots=True, 
                 figsize=(12, 10), 
                 title=f'Alphabet Inc. Stock Analysis ({start_date} to {end_date})',
                 grid=True)

plt.xlabel('Date')
plt.tight_layout()
plt.show()
```

```python
# 14. Write a Pandas program to create a plot of adjusted closing prices, thirty
# days and forty days simple moving average of Alphabet Inc. between two specific
# dates.
df['SMA30'] = df['Adj Close'].rolling(window=30).mean()
df['SMA40'] = df['Adj Close'].rolling(window=40).mean()
sma_data = df.loc[start_date:end_date, ['Adj Close', 'SMA30', 'SMA40']]
sma_data.plot(figsize=(12, 6), linewidth=2)
plt.title(f'Alphabet Inc. Adj Close with 30 & 40 Day SMA\n({start_date} to {end_date})')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend(['Adjusted Close', '30-Day SMA', '40-Day SMA'])
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# 15. Write a Pandas program to create a plot of adjusted closing prices, 30 days
# simple moving average and exponential moving average of Alphabet Inc.
# between two specific dates.
df['SMA30'] = df['Adj Close'].rolling(window=30).mean()
df['EMA30'] = df['Adj Close'].ewm(span=30, adjust=False).mean()
analysis_data = df.loc[start_date:end_date, ['Adj Close', 'SMA30', 'EMA30']]
analysis_data.plot(figsize=(8, 6), style=['-', '--', '-.'])
plt.title(f'Alphabet Inc: Adj Close, SMA30, and EMA30\n({start_date} to {end_date})')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend(['Adj Close', '30-Day SMA', '30-Day EMA'])
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# 16. Write a Pandas program to create a scatter plot of the trading volume/stock
# prices of Alphabet Inc. stock between two specific dates.
scatter_data = df.loc[start_date:end_date]
scatter_data.plot(kind='scatter', 
                  x='Volume', 
                  y='Close', 
                  figsize=(8, 6), 
                  color='darkblue', 
                  s=50,      
                  alpha=0.7)  
plt.title(f'Alphabet Inc. Scatter Plot: Volume vs. Closing Price\n({start_date} to {end_date})')
plt.xlabel('Trading Volume')
plt.ylabel('Closing Price (USD)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# 17. Write a Pandas program to create a plot to visualize daily percentage returns
# of Alphabet Inc. stock price between two specific dates.
df['Daily Return'] = df['Close'].pct_change()
returns_data = df.loc[start_date:end_date, 'Daily Return']
plt.figure(figsize=(8, 6))
returns_data.plot(color='blue', lw=1.5)
plt.title(f'Alphabet Inc. Daily Percentage Returns\n({start_date} to {end_date})')
plt.xlabel('Date')
plt.ylabel('Percentage Change')
plt.axhline(0, color='black', linestyle='--', linewidth=1, alpha=0.7)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# 18. Write a Pandas program to plot the volatility over a period of time of Alphabet
# Inc. stock price between two specific dates.
# Volatility is usually calculated as rolling standard deviation of returns
df['Daily Return'] = df['Close'].pct_change()
df['Volatility'] = df['Daily Return'].rolling(window=7).std()
volatility_data = df.loc[start_date:end_date, 'Volatility']
plt.figure(figsize=(8, 6))
volatility_data.plot(color='red', lw=2)
plt.title(f'Alphabet Inc. Stock Volatility (7-Day Rolling SD)\n({start_date} to {end_date})')
plt.xlabel('Date')
plt.ylabel('Volatility (Standard Deviation of Returns)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# 19. Write a Pandas program to create a histogram to visualize daily return
# distribution of Alphabet Inc. stock price between two specific dates.
df['Daily Return'] = df['Close'].pct_change()
returns_dist = df.loc[start_date:end_date, 'Daily Return'].dropna()
plt.figure(figsize=(8, 6))
returns_dist.plot(kind='hist', 
                  bins=30, 
                  color='seagreen', 
                  edgecolor='white', 
                  alpha=0.8)
plt.title(f'Alphabet Inc. Daily Return Distribution\n({start_date} to {end_date})')
plt.xlabel('Daily Return (Percentage Change)')
plt.ylabel('Frequency (Number of Days)')
plt.axvline(0, color='red', linestyle='--', linewidth=1)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()

# 5. Display the plot
plt.show()
```
