```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= sns.load_dataset("tips")
print(df.head())
```

```python
#1.line plot:average tip across row index-Matplotlib
plt.figure(figsize=(6,4))
plt.plot(df.index,df["tip"],marker="o")
plt.title("Line plot Tip across records")
plt.xlabel("Record index")
plt.ylabel("Tip")
plt.show()
```

```python
#1.Line plot- Seaborn
plt.figure(figsize=(6,4))
sns.lineplot(x=df.index,y=df["tip"])
plt.title("seaborn line plot-tip acorss records")
plt.xlabel("Record index")
plt.ylabel("Tip")
plt.show()
```

```python
#2.Barplot:Average total_bill for each day
avg_bill_day=df.groupby("day")["total_bill"].mean()
plt.figure(figsize=(6,4))
plt.bar(avg_bill_day.index,avg_bill_day.values)
plt.title("Bar plot-Average Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Avearge Total Bill")
plt.show()
```

```python
#2.Bar plot
plt.figure(figsize=(6,4))
sns.barplot(x="day",y="total_bill",data=df)
plt.title("seaborn bar plot-Average Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Avearge Total Bill")
plt.show()
```

```python
#3.Histogram:distribution of total_bill
plt.figure(figsize=(6,4))
plt.hist(df["total_bill"],bins=5,edgecolor= "black")
plt.title("Histogarm-Total bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
#3.histogram
plt.figure(figsize=(6,4))
sns.histplot(df["total_bill"],bins=15)
plt.title("Seaborn Histogram-Total bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
#4.Boxplot:Tip spread by day
tip_by_day=[df[df["day"]==d]["tip"] for d in df["day"].unique()]
plt.figure(figsize=(6,4))
plt.boxplot(tip_by_day,label=df["day"].unique())
plt.title("Box Plot-Tip by Day")
plt.xlabel("Day")
plt.ylabel("Tip")
plt.show()
```

```python
#4.box plot
plt.figure(figsize=(6,4))
sns.boxplot(x="day",y="tip",data=df)
plt.title("Seaborn Box Plot-Tip by Day")
plt.xlabel("Day")
plt.ylabel("Tip")
plt.show()
```

```python
#5.Scatter plot:Toatal_bil vs tip
plt.figure(figsize=(6,4))
plt.scatter(df["total_bill"],df["tip"])
plt.title("scatter plot-Toatl bill vs tip")
plt.xlabel("Toatl Bill")
plt.ylabel("Tip")
plt.show()
```

```python
#5.Scatter plot
plt.figure(figsize=(6,4))
sns.scatterplot(x="total_bill",y="tip",data=df)
plt.title("Seaborn scatter plot-Total bill vs tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
#9.Heatmap
corr=df[["total_bill","tip","size"]].corr()
plt.figure(figsize=(6,4))
# sns.heatmap(corr,annot=True,cmap="coolwarm")
sns.heatmap(corr,annot=True,cmap="crest")
plt.title("Seaborn heatmap - Correaltion matrix")
plt.show()
          
```

```python
avg_tip_by_day=df.groupby("day")["total_bill"].mean()
corr=df[["total_bill","tip","size"]].corr()
```

```python
#Traditional/Functional pyplot style
plt.figure(figsize=(30,20))
```

```python
#Line
plt.subplot(2,3,1)
plt.plot(df.index, df["total_bill"])
plt.title("Line Plot")

#Bar
plt.subplot(2,3,2)
plt.bar(avg_tip_by_day.index, avg_tip_by_day.values)
plt.title("Bar Plot")

#Histogram
plt.subplot(2,3,3)
plt.hist(df["total_bill"],bins=15)
plt.title("Histogram")

#Box
plt.subplot(2,3,4)
plt.boxplot(df["tip"])
plt.title("Box Plot")

#Scatter
plt.subplot(2,3,5)
plt.scatter(df["total_bill"],df["tip"])
plt.title("Scatter Plot")

#HeatMap
plt.subplot(2,3,6)
plt.imshow(corr)
plt.title("Heatmap")
plt.colorbar()

plt.tight_layout()
plt.show()
```

```python
#ObjectOriented Style
fig,axes=plt.subplots(2,3,figsize=(12,8))

#Line
axes[0,0].plot(df.index,df["total_bill"])
axes[0,0].set_title("Line Plot")

#Bar
axes[0,1].bar(avg_tip_by_day.index,avg_tip_by_day.values)
axes[0,1].set_title("Bar Plot")

#Histogram
axes[0,2].hist(df["total_bill"],bins=10)
axes[0,2].set_title("Hiistogram")

#Box
axes[1,0].boxplot(df["tip"])
axes[1,0].set_title("Box Plot")

#Scatter
axes[1,1].scatter(df["total_bill"],df["tip"])
axes[1,1].set_title("Scatter Plot")

#Heatmap
img=axes[1,2].imshow(corr)
axes[1,2].set_title("Heatmap")
fig.colorbar(img,ax=axes[1,2])

plt.tight_layout()
plt.show()

```
