```python
# Feature Engineering (66–75)
# 66. Create a new column `FamilySize = SibSp + Parch`.
df1['FamilySize'] = df1['SibSp'] + df1['Parch']
df1
```

```python
# 67. Create a new column `IsAlone`.
df1['IsAlone'] = np.where(df1['FamilySize'] == 0, 1, 0)
df1
```

```python
# 68. Create age groups: child, adult, senior.
df1.loc[df1["Age"]<18,"AgeGroup"]="Child"
df1.loc[df1["Age"].between(18,60),"AgeGroup"]="Adult"
df1.loc[df1["Age"]>60,"AgeGroup"]="Senior"
df1
```

```python
# 69. Create fare categories: low, medium, high.
df1.loc[df1["Fare"]<30,"FareCategory"]="low"
df1.loc[df1["Fare"].between(30,60),"FareCategory"]="medium"
df1.loc[df1["Fare"]>60,"FareCategory"]="high"
df1
```

```python
# 70. Extract title from passenger names.
df1['Title'] = df1['Name'].str.extract(' ([A-Za-z]+)\\.', expand=False)
df1[['Name', 'Title']]
```

```python
# 71. Create a column for surname.
df1['Surname'] = df1['Name'].str.split(',').str[0]
df1[['Name', 'Surname']]
```

```python
# 72. Create a column `Fare_per_person`.
df1['Family_Size'] = df1['SibSp'] + df1['Parch'] + 1
df1['Fare_per_person'] = df1['Fare'] / df1['Family_Size']
df1[['Name', 'Family_Size', 'Fare_per_person']]
```

```python
# 73. Convert `Sex` into numeric values.
df1['Sex'] = df1['Sex'].replace({'male': 0, 'female': 1})
df1[['Name', 'Sex']]
```

```python
# 74. Convert survival values into labels `Yes/No`.
df1['Survived'] = df1['Survived'].replace({0: 'No', 1: 'Yes'})
df1[['Name', 'Survived']]
```

```python
# 75. Add a column showing age after 5 years.
df1['Age_after_5'] = df1['Age'] + 5
df1
```
