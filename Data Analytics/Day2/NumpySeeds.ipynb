```python
import numpy as np
```

```python
print("Without seed:")
print("randint:", np.random.randint(1,10,5))
print("rand:", np.random.rand(5))
print("choice:", np.random.choice([1,2,3,4],5))
```

```python
print("Without seed:")
print("randint:", np.random.randint(1,10,5))
print("rand:", np.random.rand(5))
print("choice:", np.random.choice([1,2,3,4],5))
```

```python
np.random.seed(43)
print("\nWith seed(run 1):")

print("randint:",np.random.randint(1,10,5))
print("rand:",np.random.rand(5))
print("choice:",np.random.choice([1,2,3,4],5))
```

```python
np.random.seed(43)
print("\nWith seed(run 2-same output):")
print("randint:",np.random.randint(1,10,5))
print("rand:",np.random.rand(5))
print("choice:",np.random.choice([1,2,3,4],5))
```

```python
#uniform and normal
print("\nuniform:",np.random.uniform(10,50,5))
print("normal:",np.random.normal(20,5,5))
```
