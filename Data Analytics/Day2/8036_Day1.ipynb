```python
# Write a NumPy program to sort a given array by row and column in ascending order.
arr = np.array([[9, 3, 5],
                [1, 7, 2],
                [6, 4, 8]])

print("Original Array:\n", arr)
row_sorted = np.sort(arr, axis=1)
print("\nRow-wise Sorted Array:\n", row_sorted)

col_sorted = np.sort(arr, axis=0)
print("\nColumn-wise Sorted Array:\n", col_sorted)
```

```python
# Write a NumPy program to extract all numbers from a given array which are less and greater than a
# specified number.
arr = np.array([10, 25, 5, 40, 30, 15])
num = 20

less_than = arr[arr < num]
greater_than = arr[arr > num]

print("Original Array:", arr)
print("Numbers less than", num, ":", less_than)
print("Numbers greater than", num, ":", greater_than)
```

```python
# Write a NumPy program to replace all numbers in a given array which is equal, less and greater to a
# given number.
arr=np.array([10,50,25,29,58,60])
arr
num = 30
arr[arr<num]=0
arr[arr==num]=1
arr[arr>num]=-1
print("Modified Array:", arr)
```
