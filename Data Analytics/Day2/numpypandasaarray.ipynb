```python
import numpy as np
import pandas as pd
```

```python
#data
companies= np.array(["TCS","Infosys","Wipro"])
sales= np.array([100,300,200])
profit=np.array([40.80,30,55])
print("------Numpy Arrays------------")
print("companies:",companies)
print("sales:",sales)
print("profit:",profit)
```

```python
#Access using index
print("\nUsing numpy:")
print("Infosys sales:",sales[1])
print("Infosys profit:",profit[1])
```

```python
#pandas series
sales_s= pd.Series(sales,index=companies)
profit_s=pd.Series(profit,index=companies)
```

```python
print("\n\n--------pandas series-----------")
print("sales series:\n",sales_s)
print("\nprofit series:\n",profit_s)
```

```python
#access using label
print("\nUsing series:")
print("Infosys sales:",sales_s["Infosys"])
print("Infosys profit:",profit_s["Infosys"])
```

```python
df2= pd.DataFrame({
    "Employee":["Amit","Neha","Amey","Rohit","Raj"],
    "Department":["IT","HR","IT","Finance",np.nan],
    "Salary":[10000,70000,45000,76500,0]})
# print(df2)
df2
```

```python
#Indexing
df2_Indexed=df2.set_index("Employee")
# print("\n Set Employee as index:\n",df2_Indexed)
df2_Indexed
```

```python
#Filtering
# print("\nSalary>50000:\n",df2[df2["Salary"]>50000])
df2[df2["Salary"]>50000]
```

```python
#check missing
# print("\nMissing Values:\n",df2.isnull())
df2.isnull()
```

```python
#fill missing
print("\nFill misiing with:\n",df2.fillna(0))
```

```python
#Drop misiing
print("\n Drop missing rows:\n",df2.dropna())
```

```python
# Grouping
print("\n Average Salary by Department:\n",df2.groupby("Department")["Salary"].mean())
```

```python
#Merge(join)
df2_bonus=pd.DataFrame({
    "Employee":["Amit","Neha","Raj"],
    "Bonus":[5000,8000,5000]})

merged=pd.merge(df2,df2_bonus,on="Employee")
# print("\nMerged DataFrame:\n",merged)
merged
```

```python
#concat
df2_new= pd.DataFrame({
    "Employee":["Kiran"],
    "Department":["IT"],
    "Salary":[50000]})
concat_df2=pd.concat([df2,df2_new],ignore_index=True)
# print("\n Concatenated DataFrame:\n",concat_df2)
concat_df2
```
