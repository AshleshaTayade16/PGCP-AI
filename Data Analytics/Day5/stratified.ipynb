```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_2015=pd.read_csv("marathon_results_2015.csv")
df_2016=pd.read_csv("marathon_results_2016.csv")
df_2017=pd.read_csv("marathon_results_2017.csv")

df=pd.concat([df_2015,df_2016,df_2017])

original_row_count=len(df)
df
```

```python
#find the total number of male and feamle athletes
total_males=len(df[df["M/F"]=="M"])
total_females=len(df[df["M/F"]=="F"])
total_males
total_females
```

```python
#calculate the proportionate number of male and female athleths for the sample
sample_size= int(0.1*original_row_count)
sample_males= int(0.1* total_males)
sample_females= int(0.1* total_females)
print(sample_size,sample_males,sample_females)
```

```python
#Generate proportionate samples of male and feamle athleths

male_sample=df[df["M/F"]=="M"].sample(n= sample_males,random_state=42)
female_sample=df[df["M/F"]=="F"].sample(n= sample_females,random_state=42)   
print(len(male_sample),len(female_sample))
```

```python
#combine the male and feamle samples into single sample dataframe
sample_df=pd.concat([male_sample,female_sample])
```

```python
#get the total number of rows in the sampl;ed dataframe 
sample_row_count= len(sample_df)
print("First few row of the original dataframe:")
print(df.head())
print("\nFirst few row of the sampled dataframe:")
print(sample_df.head())
print(f"number of row of the original dataframe:{original_row_count}")
print(f"number of row of the sampled dataframe:{sample_row_count}")
```
