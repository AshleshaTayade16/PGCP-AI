```python
#mean,median,mode
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
temperature =[30,31,29,32,52,30,31,31,40,30,30,25]
products_sold=[20,21,30,40,20,21,30,25,40,30,20,25]

df_temp=pd.DataFrame({"Temperature":temperature})
df_sales=pd.DataFrame({"Products_sold":products_sold})
print("Temperature Data")
print("Mean:",df_temp["Temperature"].mean())
print("Median:",df_temp["Temperature"].median())
print("Mode:",df_temp["Temperature"].mode())

print("\nProducts Sold Data")
print("Mean:",df_sales["Products_sold"].mean())
print("Median:",df_sales["Products_sold"].median())
print("Mode:",df_sales["Products_sold"].mode()[0])

fig,axes =plt.subplots(1,2,figsize=(10,5))
axes[0].boxplot(df_temp["Temperature"])
axes[0].set_title("Temperature(no outlier)")
axes[0].set_ylabel("C")

axes[1].boxplot(df_sales["Products_sold"])
axes[1].set_title("Products_Sold(with outlier)")
axes[1].set_ylabel("Units")

plt.tight_layout()
plt.show()
```

```python
#Outlier Analysis
commuter_times=[16,8,35,17,13,15,15,5,16,25,20,20,12,10]

plt.figure(figsize=(10,6))
sns.boxplot(data=commuter_times,orient='h')
plt.title('Box Plot of Commuter Times')
plt.xlabel('Minutes')
plt.show()
```
