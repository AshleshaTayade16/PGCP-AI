```python
#paired T-test for t.1 and t.2
import pandas as pd
from scipy import stats
```

```python
df=pd.read_csv("hypothermia.csv")
df
```

```python
# df=df[['t.1','t.2']].dropna()
# df
```

```python
df=df[['t.1','t.2']].dropna()
sample_df=df.sample(n=30,replace=False)
t_statistic,p_value=stats.ttest_rel(sample_df['t.1'],sample_df['t.2'])
print("P-value:",p_value)
alpha=0.5
if p_value<alpha:
    print("Reject H0")
else:
    print("Do not reject H0")
```
