```python
# 48. Find value counts of a categorical column like Department.
emp=pd.DataFrame({"Name": ["Ashlesha","Ruchika","Sai","Priti","Vijay"],
                  "Dept": ["IT","Finance","IT","HR","Support"],
                  "Salary": [77000,57464,np.nan,88000,430000]})
print("Employee:",emp)
emp["Dept"].value_counts()
```
